# RD-Agent Factor Mining

This notebook demonstrates the `RDAgentRunner` for automated factor discovery
and hyperparameter optimization.

When the `rdagent` package is not installed, the runner falls back to a
Python-native simulation that mines factors from a curated template library
and persists a knowledge base to `data/rd_knowledge_base/kb.json`.


```python
import sys
from pathlib import Path

# Add project root to path
repo_root = Path('.').resolve().parent
sys.path.insert(0, str(repo_root))
print(f'Project root: {repo_root}')
```

```python
# Create RDAgentRunner and check library status
from src.core.rd_agent_runner import RDAgentRunner

runner = RDAgentRunner()

status = runner.library_status()
print('Factor Library Status:')
for key, value in status.items():
    print(f'  {key:25s}: {value}')
```

```python
# Run co-optimization: factor mining + model hyperparameter search
print('Running co_optimize (3 iterations)...')
try:
    result = runner.co_optimize(iterations=3, budget=0.5)
    print('\nco_optimize results:')
    for key, value in result.items():
        print(f'  {key:30s}: {value}')
except Exception as e:
    print(f'co_optimize error: {e}')
```

```python
# Check library status again after mining
status_after = runner.library_status()
print('Factor Library Status (after mining):')
for key, value in status_after.items():
    print(f'  {key:25s}: {value}')
```

```python
# Run library validation: 3-way comparison
print('Running validate_library...')
try:
    validation = runner.validate_library()
    print('\nValidation results:')
    for key, value in validation.items():
        print(f'  {key:25s}: {value}')
except Exception as e:
    print(f'validate_library error: {e}')
```

## Summary

`RDAgentRunner` provides:

- `co_optimize(iterations, budget)` — joint factor mining + model hyperparameter search
- `mine_factors(iterations)` — factor-only evolution
- `optimize_model(iterations)` — LightGBM hyperparameter random search
- `library_status()` — report on current factor library
- `validate_library()` — 3-way Sharpe comparison (Alpha158 vs library vs combined)
- `multi_trace_co_optimize(traces, iterations)` — parallel independent runs
- `implement_paper(source)` — derive factors from a local research paper

All factors are persisted to `outputs/factor_library.json`.
The knowledge base at `data/rd_knowledge_base/kb.json` prevents reproposals.
